```python
# Convert data types.

# Convert ISBNs to strings.
df_UK_weekly["ISBN"] = df_UK_weekly["ISBN"].astype(str)
df_ISBN["ISBN"] = df_ISBN["ISBN"].astype(str)

# Convert End_Date to datetime.
df_UK_weekly["End_Date"] = pd.to_datetime(df_UK_weekly["End_Date"])

# Check data types.
print('Weekly Trend Data (UK)','\n')
print(df_UK_weekly.info(),'\n')
print('ISBN Data (UK)','\n')
print(df_ISBN.info())
```

## Resampling Data for Time-Series:

The data was resampled to a consistent weekly frequency, ensuring that zero-sales weeks were explicitly represented for time-series modelling.

```python
# Resample weekly data by ISBN.

df_UK_weekly_resampled = (
    df_UK_weekly
    .set_index("End_Date")
    .groupby("ISBN")[numeric_cols]
    .resample("W")
    .sum()
    .fillna(0)
    .reset_index()
)

df_UK_weekly_resampled.head()
```

```python
# Add back book metadata from ISBN file.

# Keep only non-duplicated ISBN metadata.
isbn_metadata = (
    df_ISBN
    .drop_duplicates(subset="ISBN")
)

df_UK_weekly_resampled = df_UK_weekly_resampled.merge(
    isbn_metadata,
    on="ISBN",
    how="left"
)

df_UK_weekly_resampled.head()
```

```python
# Set datetime index for time-series analysis.

df_UK_weekly_resampled = (
    df_UK_weekly_resampled
    .set_index("End_Date")
    .sort_index()
)

df_UK_weekly_resampled = df_UK_weekly_resampled.rename(
    columns={"RRP_x": "RRP"}
)
df_UK_weekly_resampled = df_UK_weekly_resampled.drop(
    columns=["RRP_y"]
)

df_UK_weekly_resampled.head()
```

Check Zero-Sales Weeks After Resampling:

```python
# Check zero-sales weeks after resampling, focused on 2020–2022.

zero_check_start = "2020-01-01"
zero_check_end = "2022-12-31"

df_zero_check = df_UK_weekly_resampled.loc[
    (df_UK_weekly_resampled.index >= zero_check_start) &
    (df_UK_weekly_resampled.index <= zero_check_end)
].copy()

print("Zero-check period:")
print(df_zero_check.index.min(), "to", df_zero_check.index.max())

print("\nShape:")
print(df_zero_check.shape)
```

```python
# Summarise zero-sales weeks by date across all ISBNs.

weekly_zero_summary = (
    df_zero_check
    .assign(is_zero_volume=df_zero_check["Volume"] == 0)
    .groupby(df_zero_check.index)
    .agg(
        total_isbn_records=("ISBN", "count"),
        zero_volume_records=("is_zero_volume", "sum"),
        non_zero_volume_records=("is_zero_volume", lambda x: (~x).sum()),
        total_volume=("Volume", "sum")
    )
)

weekly_zero_summary["zero_volume_share"] = (
    weekly_zero_summary["zero_volume_records"] /
    weekly_zero_summary["total_isbn_records"]
)

display(weekly_zero_summary.head())
display(weekly_zero_summary.tail())
```

```python
# Visualise zero-sales coverage over time.

fig, axes = plt.subplots(3, 1, figsize=(14, 8), sharex=True)

weekly_zero_summary["total_isbn_records"].plot(
    ax=axes[0],
    title="Total ISBN Records by Week After Resampling"
)
axes[0].set_ylabel("ISBN Records")

weekly_zero_summary["zero_volume_records"].plot(
    ax=axes[1],
    title="Zero-Sales ISBN Records by Week"
)
axes[1].set_ylabel("Zero Records")

weekly_zero_summary["zero_volume_share"].plot(
    ax=axes[2],
    title="Share of ISBN Records with Zero Sales by Week"
)
axes[2].set_ylabel("Zero Share")

plt.tight_layout()
plt.show()
```

The dataset contains blocks of entirely missing weekly observations during 2020–2021, likely associated with a reporting interruption during the COVID period. These gaps will have downstream effects on forecast quality.

# Classical Time Series Modelling and Forecasting:

This section applies a range of classical time series techniques to understand the underlying structure of sales for The Alchemist and The Very Hungry Caterpillar.

First, the series are decomposed into trend, seasonal, and residual components to identify the most appropriate decomposition approach. Autocorrelation (ACF) and partial autocorrelation (PACF) analysis are then used to investigate temporal dependencies and inform model specification, while stationarity is assessed using formal statistical tests.

Using all observations from 1 January 2012 up to, but excluding, the final 32 weeks, Auto ARIMA is employed to identify the most suitable forecasting model for each book.

The selected models are evaluated through residual diagnostics before being used to forecast the final 32-week holdout period. Forecast performance is then assessed by comparing the predicted values and confidence intervals against the actual observed sales.

This function runs the core time-series diagnostics for a single weekly book-sales series, including trend/seasonality decomposition, autocorrelation checks, stationarity testing, and an optional Ljung-Box test to assess whether residual autocorrelation remains.

```python
# Function to handle all aspects of decomposition.

def analyse_book_series(
    series,
    book_name,
    period=52,
    use_log=False,
    acf_lags=100,
    run_ljung_box=True
):

    # Clean and optionally transform.

    y = series.dropna().copy()

    if use_log:
        y_model = np.log1p(y)
        transform_label = "log-transformed"
    else:
        y_model = y
        transform_label = "original"

    print(f"\n{'='*70}")
    print(f"{book_name}")
    print(f"Series used: {transform_label}")
    print(f"{'='*70}")

    # STL decomposition.

    stl = STL(
        y_model,
        period=period,      # 52 for weekly annual seasonality
        seasonal=7,         # controls flexibility of seasonal component
        robust=True         # reduces effect of large outliers/spikes
    )

    stl_result = stl.fit()

    fig = stl_result.plot()
    fig.set_size_inches(12, 8)
    fig.suptitle(f"STL Decomposition - {book_name}", fontsize=14)
    plt.tight_layout()
    plt.show()

    print("\nDecomposition interpretation guide:")
    print("- Trend: shows long-term movement in sales.")
    print("- Seasonal: shows recurring annual weekly pattern.")
    print("- Residual: shows unexplained variation after trend and seasonality are removed.")

    # Optional Ljung-Box test on STL residuals.

    if run_ljung_box:
        print("\nLjung-Box test on STL residuals:")
        lb_result = acorr_ljungbox(
            stl_result.resid.dropna(),
            lags=[10, 20, 52],
            return_df=True
        )
        display(lb_result)

        print("Ljung-Box interpretation:")
        print("- p < 0.05: residuals still contain autocorrelation; decomposition has not removed all structure.")
        print("- p > 0.05: no strong evidence of autocorrelation; residuals are closer to white noise.")

    # ACF and PACF plots.

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    plot_acf(y_model, lags=acf_lags, alpha=0.05, ax=axes[0])
    axes[0].set_title(f"ACF - {book_name}")

    plot_pacf(y_model, lags=acf_lags, alpha=0.05, ax=axes[1], method="ywm")
    axes[1].set_title(f"PACF - {book_name}")

    plt.tight_layout()
    plt.show()

    print("\nACF/PACF interpretation guide:")
    print("- Slow ACF decay suggests trend or non-stationarity.")
    print("- Spikes near lag 52 suggest annual seasonality in weekly data.")
    print("- PACF spikes at early lags suggest autoregressive structure.")

    # Stationarity tests.

    adf_result = adfuller(y_model)
    kpss_result = kpss(y_model, regression="c", nlags="auto")

    adf_p = adf_result[1]
    kpss_p = kpss_result[1]

    print("\nStationarity Tests")
    print(f"ADF p-value:  {adf_p:.4f}")
    print(f"KPSS p-value: {kpss_p:.4f}")

    print("\nADF interpretation:")
    if adf_p < 0.05:
        print("- Reject H0: evidence suggests the series is stationary.")
    else:
        print("- Fail to reject H0: evidence suggests the series may be non-stationary.")

    print("\nKPSS interpretation:")
    if kpss_p < 0.05:
        print("- Reject H0: evidence suggests the series is non-stationary.")
    else:
        print("- Fail to reject H0: evidence suggests the series may be stationary.")

    print("\nCombined stationarity inference:")
    if adf_p < 0.05 and kpss_p > 0.05:
        print("Likely stationary.")
    elif adf_p > 0.05 and kpss_p < 0.05:
        print("Likely non-stationary. Differencing may be required before ARIMA/SARIMA.")
    else:
        print("Mixed evidence. Inspect plots and consider first or seasonal differencing.")

    return {
        "series_used": y_model,
        "stl_result": stl_result,
        "adf_p": adf_p,
        "kpss_p": kpss_p
    }
```

## Auto ARIMA Model Selection and Forecasting:

This section applies Auto ARIMA to automatically identify the most appropriate seasonal ARIMA specification for each book using all observations from 1 January 2012 up to, but excluding, the final 32 weeks of data. The selected models are evaluated through parameter inspection and residual diagnostics before being used to forecast the final 32-week holdout period. Forecast performance is assessed by comparing predicted values and confidence intervals against the actual observed sales.

Define Relevant Functions:

```python
# Function to fit and save best Auto ARIMA model.

def fit_auto_arima_model(
    series,
    book_name,
    horizon=32,
    m=52,
    use_log=True
):
    y = series.dropna().copy().sort_index()
    y_model = np.log1p(y) if use_log else y.copy()

    train_model = y_model.iloc[:-horizon]
    train_original = y.iloc[:-horizon]
    test_original = y.iloc[-horizon:]

    print(f"\n{'='*70}")
    print(f"AUTO ARIMA MODEL FIT – {book_name.upper()}")
    print(f"{'='*70}")
    print(f"Training period : {train_original.index.min()} to {train_original.index.max()}")
    print(f"Test period     : {test_original.index.min()} to {test_original.index.max()}")
    print(f"Forecast horizon: {horizon}")
    print(f"Seasonality m   : {m}")
    print(f"Model scale     : {'log1p' if use_log else 'original'}")

    model = auto_arima(
        train_model,
        start_p=0, d=None, start_q=0,
        max_p=3, max_d=2, max_q=3,
        start_P=0, D=None, start_Q=0,
        max_P=2, max_D=1, max_Q=2,
        max_order=10,
        m=m,
        seasonal=True,
        stationary=False,
        information_criterion="aic",
        test="kpss",
        seasonal_test="ocsb",
        stepwise=True,
        suppress_warnings=True,
        error_action="ignore",
        trace=True,
        with_intercept="auto"
    )

    print(model.summary())
    print("\nBest model:")
    print("Order:", model.order)
    print("Seasonal order:", model.seasonal_order)
    print("AIC:", round(model.aic(), 2))

    return {
        "model": model,
        "train": train_original,
        "train_model": train_model,
        "test": test_original,
        "original_series": y,
        "use_log": use_log,
        "book_name": book_name,
        "horizon": horizon,
        "m": m
    }
```

```python
# Function to forecast using saved best Auto ARIMA model.

def forecast_auto_arima_results(
    arima_results,
    horizon=None,
    plot_window=150,
    y_label="Sales Volume",
    title_prefix="Auto ARIMA"
):
    model = arima_results["model"]
    train = arima_results["train"]
    train_model = arima_results["train_model"]
    test = arima_results["test"]
    y = arima_results["original_series"]
    use_log = arima_results["use_log"]
    book_name = arima_results["book_name"]

    if horizon is None:
        horizon = arima_results.get("horizon", len(test))

    forecast_model_scale, conf_int_model_scale = model.predict(
        n_periods=horizon,
        return_conf_int=True,
        alpha=0.05
    )

    forecast_model_scale = pd.Series(
        forecast_model_scale,
        index=test.index,
        name="Forecast_Model_Scale"
    )

    lower_ci_model_scale = pd.Series(
        conf_int_model_scale[:, 0],
        index=test.index,
        name="Lower_CI_Model_Scale"
    )

    upper_ci_model_scale = pd.Series(
        conf_int_model_scale[:, 1],
        index=test.index,
        name="Upper_CI_Model_Scale"
    )

    fitted_model_scale = pd.Series(
        model.predict_in_sample(),
        index=train_model.index,
        name="Fitted_Model_Scale"
    )

    if use_log:
        forecast = np.expm1(forecast_model_scale)
        lower_ci = np.expm1(lower_ci_model_scale)
        upper_ci = np.expm1(upper_ci_model_scale)
        fitted = np.expm1(fitted_model_scale)
    else:
        forecast = forecast_model_scale.copy()
        lower_ci = lower_ci_model_scale.copy()
        upper_ci = upper_ci_model_scale.copy()
        fitted = fitted_model_scale.copy()

    forecast.name = "Forecast"
    lower_ci.name = "Lower_CI"
    upper_ci.name = "Upper_CI"
    fitted.name = "Fitted"

    mae = mean_absolute_error(test, forecast)
    mape = mean_absolute_percentage_error(test, forecast)

    print(f"\n{'='*70}")
    print(f"{title_prefix.upper()} FORECAST – {book_name.upper()}")
    print(f"{'='*70}")
    print(f"MAE:  {mae:.2f}")
    print(f"MAPE: {mape * 100:.2f}%")

    forecast_preview = pd.DataFrame({
        "Actual": test,
        "Forecast": forecast,
        "Lower CI": lower_ci,
        "Upper CI": upper_ci,
        "Error": test - forecast,
        "Absolute Error": (test - forecast).abs()
    })

    print("\nForecast preview:")
    display(forecast_preview.head(8).round(2))

    forecast_summary = pd.DataFrame({
        "Metric": ["Mean", "Std", "Min", "Max"],
        "Actual": [test.mean(), test.std(), test.min(), test.max()],
        "Forecast": [forecast.mean(), forecast.std(), forecast.min(), forecast.max()]
    })

    print("\nForecast summary:")
    display(forecast_summary.round(2))

    plt.figure(figsize=(16, 5))
    y.iloc[-plot_window:].plot(label="Observed")
    fitted.iloc[-plot_window:].plot(label="Fitted Values", linestyle=":")
    forecast.plot(label="Forecast")

    plt.fill_between(
        test.index,
        lower_ci,
        upper_ci,
        alpha=0.2,
        label="95% Confidence Interval"
    )

    plt.axvline(test.index[0], linestyle="--", label="Forecast Start")
    plt.title(f"{title_prefix} Forecast – {book_name}")
    plt.xlabel("Date")
    plt.ylabel(y_label)
    plt.legend()
    plt.show()

    arima_results.update({
        "forecast_model_scale": forecast_model_scale,
        "lower_ci_model_scale": lower_ci_model_scale,
        "upper_ci_model_scale": upper_ci_model_scale,
        "fitted_model_scale": fitted_model_scale,
        "forecast": forecast,
        "lower_ci": lower_ci,
        "upper_ci": upper_ci,
        "fitted": fitted,
        "mae": mae,
        "mape": mape,
        "forecast_preview": forecast_preview,
        "forecast_summary": forecast_summary
    })

    return arima_results
```

```python
# Function to diagnose residuals.

def diagnose_auto_arima_residuals(
    arima_results,
    lags=None
):
    model = arima_results["model"]
    train_model = arima_results["train_model"]
    book_name = arima_results["book_name"]
    m = arima_results.get("m", 52)

    if lags is None:
        lags = [min(10, len(train_model) - 1), min(20, len(train_model) - 1), min(m, len(train_model) - 1)]
        lags = sorted(set([lag for lag in lags if lag > 0]))

    residuals = pd.Series(
        model.resid(),
        index=train_model.index[-len(model.resid()):],
        name="Residuals"
    ).dropna()

    ljung_box = acorr_ljungbox(
        residuals,
        lags=lags,
        return_df=True
    )

    ljung_box_clean = (
        ljung_box
        .rename(columns={
            "lb_stat": "Ljung-Box Statistic",
            "lb_pvalue": "p-value"
        })
        .round(4)
    )

    print(f"\nResidual diagnostics – {book_name}")

    print("\nResidual summary:")
    display(pd.DataFrame({
        "Mean": [residuals.mean()],
        "Std": [residuals.std()],
        "Min": [residuals.min()],
        "Max": [residuals.max()]
    }).round(4))

    print("\nLjung-Box test:")
    display(ljung_box_clean)

    fig, axes = plt.subplots(1, 3, figsize=(18, 4))

    axes[0].plot(residuals)
    axes[0].axhline(0, linestyle="--")
    axes[0].set_title("Residuals Over Time")

    axes[1].hist(residuals, bins=25)
    axes[1].set_title("Residual Distribution")

    plot_acf(
        residuals,
        lags=min(m, len(residuals) - 1),
        alpha=0.05,
        ax=axes[2]
    )
    axes[2].set_title("Residual ACF")

    fig.suptitle(f"Auto ARIMA Residual Diagnostics – {book_name}", fontsize=14)
    plt.tight_layout()
    plt.show()

    arima_results.update({
        "residuals": residuals,
        "ljung_box": ljung_box,
        "ljung_box_clean": ljung_box_clean
    })

    return arima_results
```

```python
# Function for final diagnostic summary.

def arima_model_summary(arima_results, comparison_periods=10):
    model = arima_results["model"]
    book_name = arima_results["book_name"]

    train = arima_results["train"]
    test = arima_results["test"]
    forecast = arima_results["forecast"]

    print("\n" + "=" * 70)
    print(f"AUTO ARIMA MODEL SUMMARY – {book_name.upper()}")
    print("=" * 70)

    model_summary = pd.DataFrame({
        "Metric": [
            "Order (p,d,q)",
            "Seasonal Order (P,D,Q,m)",
            "AIC"
        ],
        "Value": [
            str(model.order),
            str(model.seasonal_order),
            round(model.aic(), 2)
        ]
    })

    display(model_summary)

    n = min(
        comparison_periods,
        len(test),
        len(train)
    )

    comparison_df = pd.DataFrame({
        f"Training (Last {n})":
            train.tail(n).values,

        "Actual":
            test.head(n).values,

        "Forecast":
            forecast.head(n).values
    },
    index=test.head(n).index)

    comparison_df["Error"] = (
        comparison_df["Actual"]
        - comparison_df["Forecast"]
    )

    comparison_df["Absolute Error"] = (
        comparison_df["Error"]
        .abs()
    )

    print(f"\nForecast Comparison (First {n} Periods)")
    display(comparison_df.round(2))
```

# Machine Learning & Deep Learning Forecasting Models:

This section evaluates two advanced forecasting approaches, XGBoost and Long Short-Term Memory (LSTM) neural networks, and compares their ability to predict weekly book sales. Consistent with the earlier Auto ARIMA analysis, all observations from 1 January 2012 onwards are used, with the final 32 weeks reserved as a holdout test set for out-of-sample evaluation.

For the XGBoost model, the time series is transformed into a supervised learning problem using lagged sales observations. A forecasting pipeline is constructed and evaluated using expanding-window cross-validation, while Grid Search is used to optimise key hyperparameters, including the lag window length. The best-performing model is then used to forecast the final 32 weeks of sales for each book.

For the LSTM model, the sales series is converted into sequential input-output windows suitable for deep learning. KerasTuner is used to optimise network architecture and training parameters, and the resulting best model is used to generate forecasts for the final 32-week horizon.

Model performance is assessed using Mean Absolute Error (MAE) and Mean Absolute Percentage Error (MAPE), while visual comparisons between actual and predicted sales are used to evaluate each model’s ability to capture underlying trends, seasonality and sales volatility. The objective is to determine whether machine learning and deep learning approaches provide improvements over the classical statistical models developed in the previous section.

## XGBoost Forecasting Model - Baseline:

Train/Test Split and Lagged Sequences:

These functions prepare the time series for supervised learning: the final 32 weeks are held back as the test set, while the earlier observations are used for training. The lagged sequence function then converts the sales series into input-output pairs, where the model uses the previous lookback window of sales values to predict the next value.

```python
# Train/test split.

def prepare_forecast_split(series, horizon=32):
    y = series.dropna().copy()
    train = y.iloc[:-horizon]
    test = y.iloc[-horizon:]
    return train, test


# Create lagged input-output sequences.

def create_input_output_sequences_xgb(lookback, sequence_data):
    X, y = [], []
    sequence_data = np.asarray(sequence_data)

    for i in range(lookback, len(sequence_data)):
        X.append(sequence_data[i - lookback:i])
        y.append(sequence_data[i])

    return np.asarray(X), np.asarray(y)
```

XGBoost Training and Recursive Forecasting:

The baseline XGBoost model is trained to learn relationships between past sales patterns and the next weekly sales value. The recursive forecast then predicts one week ahead, appends that prediction to the history, and repeats the process until the full 32-week forecast horizon is produced.

```python
# Train baseline XGBoost model.

def train_xgb_baseline(X_train, y_train):
    model = XGBRegressor(
        n_estimators=400,
        max_depth=5,
        learning_rate=0.05,
        min_child_weight=1,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        random_state=seed,
        n_jobs=-1
    )

    model.fit(X_train, y_train)

    return model

# Recursive forecast.

def recursive_xgb_forecast(model, history, lookback, horizon=32):
    history = list(history)
    predictions = []

    for _ in range(horizon):
        x_input = np.asarray(history[-lookback:]).reshape(1, -1)
        pred = model.predict(x_input)[0]

        predictions.append(pred)
        history.append(pred)

    return np.asarray(predictions)
```

Full Workflow / Orchestration Layer:

This function is the orchestration layer: it runs the full XGBoost baseline workflow end-to-end, including splitting the data, creating lagged training sequences, training the model, generating recursive forecasts, calculating accuracy metrics, plotting results, and returning the key outputs.

```python
# Full baseline XGBoost workflow.

def run_xgb_baseline(series, book_name, lookback=52, horizon=32):
    y = series.dropna().copy()

    train, test = prepare_forecast_split(
        y,
        horizon=horizon
    )

    print("\n" + "=" * 70)
    print(f"XGBOOST BASELINE – {book_name.upper()}")
    print("=" * 70)
    print(f"Training period: {train.index.min()} to {train.index.max()}")
    print(f"Test period: {test.index.min()} to {test.index.max()}")
    print(f"Lookback window: {lookback} weeks")
    print(f"Forecast horizon: {horizon} weeks")

    # Create lagged supervised training data.
    X_train, y_train = create_input_output_sequences_xgb(
        lookback=lookback,
        sequence_data=train.values
    )

    print(f"\nX_train shape: {X_train.shape}")
    print(f"y_train shape: {y_train.shape}")

    # Train model.
    model = train_xgb_baseline(
        X_train,
        y_train
    )

    # Forecast final 32 weeks recursively.
    predictions = recursive_xgb_forecast(
        model=model,
        history=train.values,
        lookback=lookback,
        horizon=horizon
    )

    forecast = pd.Series(
        predictions,
        index=test.index,
        name="XGBoost Forecast"
    )

    # Metrics.
    mae = mean_absolute_error(test, forecast)
    mape = mean_absolute_percentage_error(test, forecast)

    print(f"\nForecast accuracy – {book_name}")
    print(f"MAE:  {mae:.2f}")
    print(f"MAPE: {mape * 100:.2f}%")

    preview = pd.DataFrame({
        "Actual": test,
        "Forecast": forecast,
        "Error": test - forecast,
        "Absolute Error": (test - forecast).abs()
    })

    print("\nForecast preview:")
    display(preview.head(8).round(2))

    # Plot.
    plt.figure(figsize=(14, 5))
    y.iloc[-150:].plot(label="Observed")
    forecast.plot(label="XGBoost Forecast")
    plt.axvline(test.index[0], linestyle="--", label="Forecast Start")
    plt.title(f"XGBoost Baseline Forecast – {book_name}")
    plt.xlabel("Date")
    plt.ylabel("Sales Volume")
    plt.legend()
    plt.show()

    return {
        "model": model,
        "lookback": lookback,
        "train": train,
        "test": test,
        "X_train": X_train,
        "y_train": y_train,
        "forecast": forecast,
        "mae": mae,
        "mape": mape,
        "preview": preview
    }
```

## XGBoost Forecasting Model with Hyperparamater Optimisation:

We now tune the XGBoost model using Hyperopt to search across alternative model configurations rather than relying on a manually selected baseline.

This tuning grid searches across both the time-series memory length and the main XGBoost learning/regularisation controls. Specifically, it tunes the lookback window, number of boosting trees, tree depth, learning rate, row and feature sampling rates, minimum child weight, and gamma split penalty to find the configuration with the lowest cross-validated forecast error.

```python
# Define tuning grid.

param_tuning_obj = {
    "lookback": hp.choice("lookback", [28, 52, 104]),
    "n_estimators": hp.choice("n_estimators", [200, 400, 600]),
    "max_depth": hp.choice("max_depth", [3, 5, 7]),
    "learning_rate": hp.choice("learning_rate", [0.03, 0.05, 0.1]),
    "subsample": hp.choice("subsample", [0.7, 0.8, 1.0]),
    "colsample_bytree": hp.choice("colsample_bytree", [0.7, 0.8, 1.0]),
    "min_child_weight": hp.choice("min_child_weight", [1, 3, 5]),
    "gamma": hp.choice("gamma", [0, 0.1, 0.5])
}
```

This function introduces the XGBoost tuning workflow. The first function evaluates each candidate parameter set using rolling time-series cross-validation, deliberately keeping the final 32-week test horizon untouched.

```python
# Define function for cross validation.

def cross_validate_xgb_params(
    series,
    params,
    horizon=32,
    initial_ratio=0.7,
    step_size=26
):
    y = series.dropna().copy()

    # Keep final 32 weeks untouched.
    train_full = y.iloc[:-horizon]

    lookback = int(params["lookback"])
    initial_window = int(len(train_full) * initial_ratio)

    cv_scores = []

    for split_end in range(initial_window, len(train_full) - horizon + 1, step_size):

        cv_train = train_full.iloc[:split_end]
        cv_val = train_full.iloc[split_end:split_end + horizon]

        if len(cv_train) <= lookback:
            continue

        X_train, y_train = create_input_output_sequences_xgb(
            lookback=lookback,
            sequence_data=cv_train.values
        )

        model = XGBRegressor(
            n_estimators=int(params["n_estimators"]),
            max_depth=int(params["max_depth"]),
            learning_rate=float(params["learning_rate"]),
            subsample=float(params["subsample"]),
            colsample_bytree=float(params["colsample_bytree"]),
            min_child_weight=int(params["min_child_weight"]),
            gamma=float(params["gamma"]),
            objective="reg:squarederror",
            random_state=seed,
            n_jobs=-1
        )

        model.fit(X_train, y_train)

        predictions = recursive_xgb_forecast(
            model=model,
            history=cv_train.values,
            lookback=lookback,
            horizon=len(cv_val)
        )

        mae = mean_absolute_error(cv_val, predictions)
        mape = mean_absolute_percentage_error(cv_val, predictions)

        cv_scores.append({
            "train_end": cv_train.index[-1],
            "validation_start": cv_val.index[0],
            "validation_end": cv_val.index[-1],
            "mae": mae,
            "mape": mape
        })

    cv_scores = pd.DataFrame(cv_scores)

    return cv_scores
```

The second function converts that cross-validation process into a Hyperopt objective, using average MAE as the loss function to minimise.

```python
def xgb_tuning_objective_cv(params, series, horizon=32):
    cv_scores = cross_validate_xgb_params(
        series=series,
        params=params,
        horizon=horizon,
        initial_ratio=0.7,
        step_size=26
    )

    if cv_scores.empty:
        return {
            "loss": np.inf,
            "status": STATUS_OK,
            "mae": np.inf,
            "mape": np.inf,
            "params": params,
            "cv_scores": cv_scores
        }

    mean_mae = cv_scores["mae"].mean()
    mean_mape = cv_scores["mape"].mean()

    return {
        "loss": mean_mae,
        "status": STATUS_OK,
        "mae": mean_mae,
        "mape": mean_mape,
        "params": params,
        "cv_scores": cv_scores
    }
```

The third function runs the Hyperopt search across the defined parameter space, stores the results from each trial, and ranks the configurations by cross-validated MAE.

```python
def tune_xgb_model_cv(series, book_name, horizon=32, max_evals=50):
    trials = Trials()

    objective = lambda params: xgb_tuning_objective_cv(
        params=params,
        series=series,
        horizon=horizon
    )

    best_hyperparams = fmin(
        fn=objective,
        space=param_tuning_obj,
        algo=tpe.suggest,
        max_evals=max_evals,
        trials=trials,
        rstate=np.random.default_rng(seed)
    )

    trial_results = pd.DataFrame([
        {
            "cv_mae": trial["result"]["mae"],
            "cv_mape": trial["result"]["mape"],
            **trial["result"]["params"]
        }
        for trial in trials.trials
    ]).sort_values("cv_mae").reset_index(drop=True)

    print(f"\nBest CV hyperparameters – {book_name}")
    display(trial_results.head(10).round(4))

    return {
        "best_hyperparams": best_hyperparams,
        "trials": trials,
        "trial_results": trial_results
    }
```

The final function applies the best selected parameters, retrains XGBoost on the full training period, and evaluates the tuned model on the untouched 32-week holdout set.

```python
# Function to run XGBoost with best parameters.

def run_tuned_xgb_model(series, book_name, tuning_results, horizon=32):
    best = tuning_results["trial_results"].iloc[0]

    lookback = int(best["lookback"])

    y = series.dropna().copy()
    train = y.iloc[:-horizon]
    test = y.iloc[-horizon:]

    X_train, y_train = create_input_output_sequences_xgb(
        lookback=lookback,
        sequence_data=train.values
    )

    model = XGBRegressor(
        n_estimators=int(best["n_estimators"]),
        max_depth=int(best["max_depth"]),
        learning_rate=float(best["learning_rate"]),
        subsample=float(best["subsample"]),
        colsample_bytree=float(best["colsample_bytree"]),
        min_child_weight=int(best["min_child_weight"]),
        gamma=float(best["gamma"]),
        objective="reg:squarederror",
        random_state=seed,
        n_jobs=-1
    )

    model.fit(X_train, y_train)

    predictions = recursive_xgb_forecast(
        model=model,
        history=train.values,
        lookback=lookback,
        horizon=horizon
    )

    forecast = pd.Series(
        predictions,
        index=test.index,
        name="Tuned XGBoost Forecast"
    )

    mae = mean_absolute_error(test, forecast)
    mape = mean_absolute_percentage_error(test, forecast)

    print("\n" + "=" * 70)
    print(f"TUNED XGBOOST MODEL – {book_name.upper()}")
    print("=" * 70)

    print("\nBest CV-selected parameters:")
    display(best.to_frame().T.round(4))

    print(f"\nFinal holdout performance:")
    print(f"MAE:  {mae:.2f}")
    print(f"MAPE: {mape * 100:.2f}%")

    preview = pd.DataFrame({
        "Actual": test,
        "Forecast": forecast,
        "Error": test - forecast,
        "Absolute Error": (test - forecast).abs()
    })

    print("\nForecast preview:")
    display(preview.head(8).round(2))

    plt.figure(figsize=(14, 5))
    y.iloc[-150:].plot(label="Observed")
    forecast.plot(label="Tuned XGBoost Forecast")
    plt.axvline(test.index[0], linestyle="--", label="Forecast Start")
    plt.title(f"Tuned XGBoost Forecast – {book_name}")
    plt.xlabel("Date")
    plt.ylabel("Sales Volume")
    plt.legend()
    plt.show()

    return {
        "model": model,
        "best_params": best,
        "lookback": lookback,
        "train": train,
        "test": test,
        "forecast": forecast,
        "mae": mae,
        "mape": mape,
        "preview": preview
    }
```

## XGBoost Forecasting Model using sktime Pipeline:

This section refactors the XGBoost model to use the sktime pipeline. The TransformedTargetForecaster handles the deseasonalizing, detrending, and lagged sequence generation.

This function builds an enhanced XGBoost forecasting pipeline that first removes annual seasonality, then removes a simple linear trend, before training XGBoost on lagged weekly sales values.

```python
# Build sktime XGBoost pipeline.

def build_sktime_xgb_pipeline(
    window_length=52,
    xgb_params=None
):
    if xgb_params is None:
        xgb_params = {
            "n_estimators": 400,
            "max_depth": 5,
            "learning_rate": 0.05
        }

    regressor = XGBRegressor(
        objective="reg:squarederror",
        random_state=seed,
        n_jobs=-1,
        **xgb_params
    )

    pipeline = TransformedTargetForecaster([
        (
            "deseasonalize",
            Deseasonalizer(
                sp=52,
                model="additive"
            )
        ),
        (
            "detrend",
            Detrender(
                forecaster=PolynomialTrendForecaster(
                    degree=1
                )
            )
        ),
        (
            "forecast",
            make_reduction(
                regressor,
                window_length=window_length,
                strategy="recursive"
            )
        )
    ])

    return pipeline
```

The workflow then applies this pipeline to a book-sales series by preparing the weekly index, splitting the data into training and 32-week holdout periods, fitting the model, generating forecasts, calculating MAE and MAPE, and plotting the results.

```python
# Run sktime XGBoost pipeline.

def run_sktime_xgb_pipeline(
    series,
    book_name,
    window_length=52,
    horizon=32,
    xgb_params=None
):
    y = series.dropna().copy()

    y.index = y.index.to_period("W")

    train = y.iloc[:-horizon]
    test = y.iloc[-horizon:]

    print("\n" + "=" * 70)
    print(f"SKTIME XGBOOST PIPELINE – {book_name.upper()}")
    print("=" * 70)

    pipeline = build_sktime_xgb_pipeline(
        window_length=window_length,
        xgb_params=xgb_params
    )

    pipeline.fit(train)

    fh = np.arange(1, horizon + 1)

    forecast = pipeline.predict(fh=fh)

    forecast.index = test.index
    forecast.name = "sktime XGBoost Forecast"

    mae = mean_absolute_error(test, forecast)
    mape = mean_absolute_percentage_error(test, forecast)

    print("\nPipeline configuration:")

    config = pd.DataFrame({
        "Parameter": [
            "Window Length",
            "Seasonality",
            "Trend Degree"
        ],
        "Value": [
            window_length,
            52,
            1
        ]
    })

    display(config)

    print("\nFinal holdout performance:")

    print(f"MAE:  {mae:.2f}")
    print(f"MAPE: {mape * 100:.2f}%")

    preview = pd.DataFrame({
        "Actual": test.astype(float),
        "Forecast": forecast.astype(float),
        "Error": test - forecast,
        "Absolute Error": (test - forecast).abs()
    })

    print("\nForecast preview:")

    display(
        preview
        .head(8)
        .round(2)
    )

    # Plot.
    plt.figure(figsize=(14, 5))

    y_plot = y.copy()
    y_plot.index = y_plot.index.to_timestamp()

    test_plot = test.copy()
    test_plot.index = test_plot.index.to_timestamp()

    forecast_plot = forecast.copy()
    forecast_plot.index = forecast_plot.index.to_timestamp()

    y_plot.iloc[-150:].plot(
        label="Observed"
    )

    forecast_plot.plot(
        label="sktime XGBoost Forecast"
    )

    plt.axvline(
        test_plot.index[0],
        linestyle="--",
        label="Forecast Start"
    )

    plt.title(
        f"sktime XGBoost Forecast – {book_name}"
    )

    plt.xlabel("Date")
    plt.ylabel("Sales Volume")
    plt.legend()
    plt.show()

    return {
        "pipeline": pipeline,
        "train": train,
        "test": test,
        "forecast": forecast,
        "mae": mae,
        "mape": mape,
        "preview": preview,
        "window_length": window_length,
        "xgb_params": xgb_params
    }
```

## LSTM Model - Baseline:

This stage introduces an LSTM model as a neural-network forecasting baseline. Unlike ARIMA, which focuses on linear time-series structure, and XGBoost, which learns from fixed lagged features, an LSTM is designed to learn sequential patterns directly from ordered historical sales data.

The sales series is first converted into rolling input-output sequences, where the model uses the previous 28 weeks of scaled sales values to predict the next week. The final 32 weeks are held out as the test set, and scaling is fitted only on the training data to avoid leakage.

The baseline LSTM uses a simple architecture: one LSTM layer with 32 units, followed by dropout to reduce overfitting, and a final dense layer that outputs the next forecast value. This provides a neural benchmark for testing whether a sequence model can better capture nonlinear sales dynamics, short-term momentum, and irregular demand patterns.

```python
# Create LSTM input-output sequences.

def create_lstm_sequences(values, lookback=28):
    X, y = [], []

    for i in range(lookback, len(values)):
        X.append(values[i - lookback:i, 0])
        y.append(values[i, 0])

    X = np.array(X).reshape(-1, lookback, 1)
    y = np.array(y)

    return X, y
```

```python
# Prepare train/test data for LSTM.

def prepare_lstm_data(series, horizon=32, lookback=28):
    y = series.dropna().copy()

    # Hold out final 32 weeks.
    train = y.iloc[:-horizon]
    test = y.iloc[-horizon:]

    # Fit scaler on training data only.
    scaler = MinMaxScaler()
    train_scaled = scaler.fit_transform(
        train.values.reshape(-1, 1)
    )

    # Create supervised LSTM sequences from training data.
    X_train, y_train = create_lstm_sequences(
        train_scaled,
        lookback=lookback
    )

    return train, test, scaler, X_train, y_train
```

This function builds a shallow LSTM as a baseline model.

```python
# Baseline LSTM model builder.

def build_baseline_lstm_model(lookback=28):
    model = Sequential([
        LSTM(
            units=32,
            input_shape=(lookback, 1)
        ),
        Dropout(0.2),
        Dense(1)
    ])

    model.compile(
        optimizer="adam",
        loss="mean_squared_error"
    )

    return model
```

The recursive LSTM forecast works in the same broad way as the XGBoost recursive forecast: it predicts one week ahead, appends that prediction back into the input history, and repeats the process until the full forecast horizon is produced. The key difference is that the LSTM input must be reshaped into a three-dimensional sequence format because LSTMs require data as samples × timesteps × features, and the predictions must then be inverse-transformed because the model is trained on scaled values rather than original sales units.

```python
# Recursive LSTM forecast.

def recursive_lstm_forecast(model, train, scaler, lookback=28, horizon=32):
    history = list(train.values)
    predictions = []

    for _ in range(horizon):

        # Scale current history using scaler fitted on training data only.
        history_scaled = scaler.transform(
            np.array(history).reshape(-1, 1)
        ).flatten()

        # Use the most recent lookback window.
        x_input = np.array(
            history_scaled[-lookback:]
        ).reshape(1, lookback, 1)

        # Predict next value on scaled scale.
        pred_scaled = model.predict(
            x_input,
            verbose=0
        )[0, 0]

        # Convert prediction back to sales units.
        pred = scaler.inverse_transform(
            np.array([[pred_scaled]])
        )[0, 0]

        predictions.append(pred)

        # Append original-scale prediction to history.
        history.append(pred)

    return np.array(predictions)
```

```python
# Evaluate LSTM forecast.

def evaluate_lstm_forecast(test, predictions, book_name):
    forecast = pd.Series(
        predictions,
        index=test.index,
        name="LSTM Forecast"
    )

    mae = mean_absolute_error(test, forecast)
    mape = mean_absolute_percentage_error(test, forecast)

    print(f"\nLSTM Forecast Accuracy – {book_name}")
    print(f"MAE:  {mae:.2f}")
    print(f"MAPE: {mape * 100:.2f}%")

    preview = pd.DataFrame({
        "Actual": test,
        "Forecast": forecast,
        "Error": test - forecast,
        "Absolute Error": (test - forecast).abs()
    })

    print("\nForecast preview:")
    display(preview.head(8).round(2))

    return {
        "forecast": forecast,
        "mae": mae,
        "mape": mape,
        "preview": preview
    }
```

This function acts as the orchestration layer for the LSTM forecast workflow, generating recursive predictions, evaluating forecast accuracy, plotting the result, and returning the fitted model outputs in a structured results dictionary.

```python
# Full baseline LSTM forecast workflow.

def run_lstm_baseline_forecast(
    model,
    series,
    train,
    test,
    scaler,
    book_name,
    lookback=28,
    horizon=32
):
    predictions = recursive_lstm_forecast(
        model=model,
        train=train,
        scaler=scaler,
        lookback=lookback,
        horizon=horizon
    )

    evaluation = evaluate_lstm_forecast(
        test=test,
        predictions=predictions,
        book_name=book_name
    )

    plot_lstm_forecast(
        series=series,
        test=test,
        forecast=evaluation["forecast"],
        book_name=book_name
    )

    return {
        "model": model,
        "train": train,
        "test": test,
        "scaler": scaler,
        "forecast": evaluation["forecast"],
        "mae": evaluation["mae"],
        "mape": evaluation["mape"],
        "preview": evaluation["preview"]
    }
```

## Building a Stacked LSTM:

```python
# Function for Simpler Stacked LSTM.

def build_stacked_lstm_model_v1(lookback=28):
    model = Sequential([
        LSTM(
            units=64,
            return_sequences=True,
            input_shape=(lookback, 1)
        ),
        Dropout(0.2),

        LSTM(
            units=32
        ),
        Dropout(0.2),

        Dense(1)
    ])

    model.compile(
        optimizer="adam",
        loss="mean_squared_error"
    )

    return model
```

```python
# Function for Deeper stacked LSTM.

def build_stacked_lstm_model_v2(lookback=28):
    model = Sequential([
        LSTM(
            units=128,
            return_sequences=True,
            input_shape=(lookback, 1)
        ),
        Dropout(0.3),

        LSTM(
            units=64,
            return_sequences=True
        ),
        Dropout(0.3),

        LSTM(
            units=32
        ),
        Dropout(0.2),

        Dense(16, activation="relu"),
        Dense(1)
    ])

    model.compile(
        optimizer="adam",
        loss="mean_squared_error"
    )

    return model
```

```python
# Generic training function.

def train_lstm_model(model, X_train, y_train, book_name, model_name):
    print("\n" + "=" * 70)
    print(f"{model_name.upper()} – {book_name.upper()}")
    print("=" * 70)

    history = model.fit(
        X_train,
        y_train,
        epochs=75,
        batch_size=32,
        validation_split=0.2,
        verbose=1
    )

    return model, history
```

## Building a Stacked LSTM with Early Stopping:

Early stopping is used to reduce overfitting by monitoring validation loss during training and stopping once the model stops improving. This allows the stacked LSTM to train for long enough to learn useful sequential patterns, while avoiding unnecessary epochs that may cause it to memorise the training data rather than generalise to the holdout forecast period.

```python
# Build stacked LSTM with early stopping.

def train_lstm_model_with_early_stopping(
    model,
    X_train,
    y_train,
    book_name,
    model_name,
    epochs=150,
    batch_size=32
):
    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    )

    print("\n" + "=" * 70)
    print(f"{model_name.upper()} – {book_name.upper()}")
    print("=" * 70)

    history = model.fit(
        X_train,
        y_train,
        epochs=epochs,
        batch_size=batch_size,
        validation_split=0.2,
        callbacks=[early_stop],
        verbose=1
    )

    return model, history
```

## Optimising an LSTM with Keras Tuner:

Keras Tuner is a hyperparameter optimisation library for TensorFlow/Keras models that automatically tests different neural-network configurations. Here, it is used to optimise the LSTM by searching over choices such as the number of LSTM units, whether to add a second LSTM layer, dropout rates, dense-layer structure, and learning rate. The aim is to find the architecture that minimises validation loss, rather than relying on a manually chosen baseline design.

```python
# Implementing Customisable Model Function.

def build_tuned_lstm_model(hp, lookback=28):
    model = Sequential()

    # First LSTM layer.
    model.add(
        LSTM(
            units=hp.Choice("lstm_units_1", [32, 64, 128]),
            return_sequences=hp.Boolean("use_second_lstm"),
            input_shape=(lookback, 1)
        )
    )

    model.add(
        Dropout(
            hp.Choice("dropout_1", [0.1, 0.2, 0.3, 0.4])
        )
    )

    # Optional second LSTM layer.
    if hp.get("use_second_lstm"):
        model.add(
            LSTM(
                units=hp.Choice("lstm_units_2", [16, 32, 64]),
                return_sequences=False
            )
        )

        model.add(
            Dropout(
                hp.Choice("dropout_2", [0.1, 0.2, 0.3])
            )
        )

    # Optional dense layer.
    if hp.Boolean("use_dense_layer"):
        model.add(
            Dense(
                units=hp.Choice("dense_units", [8, 16, 32]),
                activation="relu"
            )
        )

    model.add(Dense(1))

    learning_rate = hp.Choice(
        "learning_rate",
        [0.001, 0.0005, 0.0001]
    )

    model.compile(
        optimizer=Adam(learning_rate=learning_rate),
        loss="mean_squared_error"
    )

    return model
```

```python
# Implementing Keras Tuner.

def tune_lstm_model(
    X_train,
    y_train,
    book_name,
    lookback=28,
    max_trials=15
):
    tuner = kt.RandomSearch(
        hypermodel=lambda hp: build_tuned_lstm_model(
            hp,
            lookback=lookback
        ),
        objective="val_loss",
        max_trials=max_trials,
        executions_per_trial=1,
        overwrite=True,
        directory="lstm_tuning",
        project_name=book_name.replace(" ", "_").lower()
    )

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    )

    tuner.search(
        X_train,
        y_train,
        epochs=100,
        batch_size=32,
        validation_split=0.2,
        callbacks=[early_stop],
        verbose=1
    )

    best_model = tuner.get_best_models(num_models=1)[0]
    best_params = tuner.get_best_hyperparameters(num_trials=1)[0].values

    print("\nBest LSTM hyperparameters:")
    display(pd.DataFrame(
        list(best_params.items()),
        columns=["Parameter", "Value"]
    ))

    return best_model, best_params, tuner
```

# Hybrid Forecasting Models:

This section evaluates whether combining classical statistical forecasting with deep learning can improve predictive performance relative to the standalone models developed previously. Two hybrid approaches are considered.

First, a sequential SARIMA–LSTM hybrid is constructed, where the SARIMA model captures the linear trend and seasonal structure of the series, and an LSTM model is trained on the SARIMA residuals to model any remaining non-linear patterns.

Second, a parallel SARIMA–LSTM hybrid is developed, where forecasts from the two models are combined using a weighted-average ensemble. For both approaches, all observations from 1 January 2012 up to, but excluding, the final 32 weeks are used for training, with the final 32 weeks retained as an out-of-sample test set.

Model performance is assessed using Mean Absolute Error (MAE) and Mean Absolute Percentage Error (MAPE), and the relative effectiveness of different ensemble weighting schemes is evaluated.

## Building a sequential SARIMA–LSTM hybrid:

```python
# Prepare series for decomposition / SARIMA.

def prepare_series_for_sarima(series, book_name=None, freq="W"):
    y = series.dropna().copy()
    y = y.sort_index()

    # Ensure numeric.
    y = pd.to_numeric(y, errors="coerce").dropna()

    # Try to enforce weekly frequency.
    y = y.asfreq(freq)

    # Fill any gaps created by asfreq.
    y = y.interpolate(method="time")

    if book_name is not None:
        print("=" * 70)
        print(f"SARIMA DATA PREPARATION – {book_name.upper()}")
        print("=" * 70)
        print(f"Observations : {len(y)}")
        print(f"Start date   : {y.index.min()}")
        print(f"End date     : {y.index.max()}")
        print(f"Missing vals : {y.isna().sum()}")
        print(f"Min value    : {y.min():.2f}")
        print(f"Max value    : {y.max():.2f}")

    return y
```

```python
# Decompose weekly sales series.

def decompose_book_series(series, book_name, model="additive", period=52):
    y = prepare_series_for_sarima(
        series,
        book_name=book_name,
        freq="W"
    )

    decomposition = seasonal_decompose(
        y,
        model=model,
        period=period
    )

    fig = decomposition.plot()
    fig.set_size_inches(14, 8)
    fig.suptitle(f"Seasonal Decomposition – {book_name}", y=1.02)
    plt.show()

    return {
        "series": y,
        "decomposition": decomposition,
        "trend": decomposition.trend,
        "seasonal": decomposition.seasonal,
        "resid": decomposition.resid
    }
```

```python
# Apply decomposition.

vhc_decomposition = decompose_book_series(
    very_hungry_caterpillar_series,
    "The Very Hungry Caterpillar",
    model="additive",
    period=52
)

alchemist_decomposition = decompose_book_series(
    alchemist_series,
    "The Alchemist",
    model="additive",
    period=52
)
```

This SARIMA function follows the same broad structure as the earlier Auto ARIMA workflow: it prepares the series, optionally applies a log transform, holds back the final forecast horizon as an untouched test set, fits a seasonal ARIMA model on training data only, reports the selected order/seasonal order/AIC, and returns the fitted model in a reusable results dictionary.

The main difference is that this version is designed specifically for the hybrid SARIMA-LSTM workflow, so it also calculates in-sample fitted values and residuals immediately after fitting. Those residuals are then used as the target series for the residual LSTM, allowing the hybrid model to learn whatever structure SARIMA failed to capture.

It is also slightly more flexible than the earlier Auto ARIMA function because it can either search for the best SARIMA specification using auto_arima, or accept predefined order and seasonal_order values. In short, the earlier Auto ARIMA functions are mainly built for standalone forecasting and evaluation, while this SARIMA function is built to produce both a classical forecast base and residuals for the sequential hybrid model.

```python
# Function for Fitting SARIMA and calculating residuals.

def fit_sarima_model(series, book_name, horizon=32, m=52, use_log=True, order=None, seasonal_order=None):
    y = prepare_series_for_sarima(
        series,
        book_name=book_name,
        freq="W"
    )

    y_model = np.log1p(y) if use_log else y.copy()

    train_model = y_model.iloc[:-horizon]
    train = y.iloc[:-horizon]
    test = y.iloc[-horizon:]

    print("\n" + "=" * 70)
    print(f"SARIMA MODEL FIT – {book_name.upper()}")
    print("=" * 70)
    print(f"Training period : {train.index.min()} to {train.index.max()}")
    print(f"Test period     : {test.index.min()} to {test.index.max()}")

    if order is not None and seasonal_order is not None:
        print(f"\nFitting SARIMA with predefined parameters: Order {order}, Seasonal Order {seasonal_order}")
        model = pm.ARIMA(order=order, seasonal_order=seasonal_order, suppress_warnings=True)
        model.fit(train_model)
    else:
        print("\nSearching for optimal parameters using auto_arima...")
        model = pm.auto_arima(
            train_model,
            seasonal=True,
            m=m,
            start_p=0,
            start_q=0,
            max_p=3,
            max_q=3,
            max_d=2,
            start_P=0,
            start_Q=0,
            max_P=2,
            max_Q=2,
            max_D=1,
            test="kpss",
            seasonal_test="ocsb",
            information_criterion="aic",
            stepwise=True,
            suppress_warnings=True,
            error_action="ignore",
            trace=True
        )

    print("\nSelected SARIMA model:")
    print("Order:", model.order)
    print("Seasonal order:", model.seasonal_order)
    print("AIC:", model.aic())

    fitted_vals = model.predict_in_sample()
    fitted_vals = np.expm1(fitted_vals) if use_log else fitted_vals

    fitted = pd.Series(
        fitted_vals,
        index=train.index,
        name="SARIMA Fitted"
    )

    residuals = train - fitted
    residuals.name = "SARIMA Residuals"

    return {
        "series": y,
        "train": train,
        "test": test,
        "model": model,
        "fitted": fitted,
        "residuals": residuals,
        "use_log": use_log
    }
```

This function converts the SARIMA residuals into supervised LSTM training data, so the LSTM can learn patterns left unexplained by the SARIMA model. It scales the residuals, creates lookback-based input-output sequences, and returns the prepared residual dataset for training the hybrid model.

```python
# Prepare SARIMA residuals for LSTM.

def prepare_lstm_residual_data(sarima_results, lookback=52):
    residuals = sarima_results["residuals"].dropna().copy()

    scaler = MinMaxScaler()
    residuals_scaled = scaler.fit_transform(
        residuals.values.reshape(-1, 1)
    )

    X_train, y_train = create_lstm_sequences(
        residuals_scaled,
        lookback=lookback
    )

    print("=" * 60)
    print("LSTM RESIDUAL DATA PREPARATION")
    print("=" * 60)
    print(f"Residual observations : {len(residuals)}")
    print(f"Lookback window       : {lookback}")
    print(f"X_train shape         : {X_train.shape}")
    print(f"y_train shape         : {y_train.shape}")

    return residuals, scaler, X_train, y_train
```

```python
# Get residuals.

vhc_residuals, vhc_resid_scaler, X_vhc_resid_lstm, y_vhc_resid_lstm = prepare_lstm_residual_data(
    vhc_sarima_results,
    lookback=52
)

alchemist_residuals, alchemist_resid_scaler, X_alchemist_resid_lstm, y_alchemist_resid_lstm = prepare_lstm_residual_data(
    alchemist_sarima_results,
    lookback=52
)
```

This residual LSTM is a deeper stacked LSTM designed to model the patterns left over after SARIMA has already captured the main trend, seasonality, and autocorrelation in the original sales series. It uses three LSTM layers with decreasing units (128 → 64 → 32) so the model can learn residual structure at different levels of abstraction, while dropout and a small dense layer help reduce overfitting and translate the learned sequence representation into a one-step residual forecast.

The lookback is 13 weeks because the LSTM is only modelling SARIMA residuals, not the full sales series. A shorter quarter-length window is enough to capture recent residual momentum or local error patterns, while avoiding the risk of overfitting noisy residuals with a full 52-week history that SARIMA has already partly explained.

```python
# Build deeper stacked LSTM model for SARIMA residuals.

def build_residual_lstm_model(lookback=13):
    model = Sequential([
        LSTM(
            units=128,
            return_sequences=True,
            input_shape=(lookback, 1)
        ),
        Dropout(0.3),

        LSTM(
            units=64,
            return_sequences=True
        ),
        Dropout(0.3),

        LSTM(
            units=32
        ),
        Dropout(0.2),

        Dense(16, activation="relu"),
        Dense(1)
    ])

    model.compile(
        optimizer="adam",
        loss="mean_squared_error"
    )

    return model
```

The residual LSTM is trained on SARIMA errors rather than raw sales, with early stopping used to stop training once the residual-learning process no longer improves.

```python
# Train residual LSTM model.

def train_residual_lstm_model(X_train, y_train, lookback=13, epochs=50, batch_size=16):
    model = build_residual_lstm_model(
        lookback=lookback
    )

    early_stop = EarlyStopping(
        monitor="loss",
        patience=8,
        restore_best_weights=True
    )

    history = model.fit(
        X_train,
        y_train,
        epochs=epochs,
        batch_size=batch_size,
        callbacks=[early_stop],
        verbose=0
    )

    return model, history
```

Future SARIMA residuals are forecast recursively by using the most recent residual history, predicting the next residual, appending it back into the sequence, and repeating this across the 32-week horizon.

```python
# Recursively forecast SARIMA residuals.

def recursive_lstm_residual_forecast(model, residuals, scaler, lookback=13, horizon=32):
    residual_history = residuals.values.reshape(-1, 1)

    residual_history_scaled = scaler.transform(
        residual_history
    ).flatten().tolist()

    scaled_predictions = []

    for _ in range(horizon):
        x_input = np.array(
            residual_history_scaled[-lookback:]
        ).reshape(1, lookback, 1)

        pred_scaled = model.predict(
            x_input,
            verbose=0
        )[0, 0]

        scaled_predictions.append(pred_scaled)
        residual_history_scaled.append(pred_scaled)

    residual_predictions = scaler.inverse_transform(
        np.array(scaled_predictions).reshape(-1, 1)
    ).flatten()

    return residual_predictions
```

The final sequential hybrid forecast is reconstructed by adding the SARIMA forecast to the LSTM-predicted residual forecast, so the classical model captures the main time-series structure while the LSTM attempts to correct its remaining errors.

```python
# Reconstruct sequential hybrid forecast.

def reconstruct_sequential_hybrid_forecast(
    sarima_results,
    residual_lstm_model,
    residual_scaler,
    residual_lookback=13,
    horizon=32
):
    test = sarima_results["test"]
    residuals = sarima_results["residuals"]

    sarima_forecast = sarima_results["model"].predict(
        n_periods=horizon
    )

    # Inverse log transformation if the model was trained on logged data.
    if sarima_results.get("use_log", False):
        sarima_forecast = np.expm1(sarima_forecast)

    sarima_forecast = pd.Series(
        sarima_forecast.values,
        index=test.index,
        name="SARIMA Forecast"
    )

    residual_forecast = recursive_lstm_residual_forecast(
        model=residual_lstm_model,
        residuals=residuals,
        scaler=residual_scaler,
        lookback=residual_lookback,
        horizon=horizon
    )

    residual_forecast = pd.Series(
        residual_forecast,
        index=test.index,
        name="LSTM Residual Forecast"
    )

    hybrid_forecast = sarima_forecast + residual_forecast
    hybrid_forecast.name = "Sequential Hybrid Forecast"

    return sarima_forecast, residual_forecast, hybrid_forecast
```

```python
# Create confidence intervals in plots.

def create_hybrid_confidence_interval(
    forecast,
    residuals,
    z_score=1.96
):
    residual_std = residuals.dropna().std()

    lower = forecast - z_score * residual_std
    upper = forecast + z_score * residual_std

    confidence_interval = pd.DataFrame({
        "Lower CI": lower.clip(lower=0),
        "Upper CI": upper
    })

    return confidence_interval
```

```python
# Evaluate and plot sequential hybrid forecast.

def evaluate_sequential_hybrid(
    sarima_results,
    sarima_forecast,
    residual_forecast,
    hybrid_forecast,
    book_name
):
    test = sarima_results["test"]
    series = sarima_results["series"]

    mae = mean_absolute_error(test, hybrid_forecast)
    mape = mean_absolute_percentage_error(test, hybrid_forecast)

    confidence_interval = create_hybrid_confidence_interval(
        forecast=hybrid_forecast,
        residuals=test - hybrid_forecast
    )

    print("\n" + "=" * 70)
    print(f"SEQUENTIAL SARIMA + LSTM HYBRID – {book_name.upper()}")
    print("=" * 70)
    print(f"MAE:  {mae:.2f}")
    print(f"MAPE: {mape * 100:.2f}%")

    preview = pd.DataFrame({
        "Actual": test,
        "SARIMA Forecast": sarima_forecast,
        "LSTM Residual Forecast": residual_forecast,
        "Hybrid Forecast": hybrid_forecast,
        "Lower CI": confidence_interval["Lower CI"],
        "Upper CI": confidence_interval["Upper CI"],
        "Hybrid Error": test - hybrid_forecast,
        "Absolute Error": (test - hybrid_forecast).abs()
    })

    display(preview.head(8).round(2))

    plt.figure(figsize=(14, 5))
    series.iloc[-150:].plot(label="Observed")
    sarima_forecast.plot(label="SARIMA Forecast", linestyle="--")
    hybrid_forecast.plot(label="Sequential Hybrid Forecast")

    plt.fill_between(
        confidence_interval.index,
        confidence_interval["Lower CI"],
        confidence_interval["Upper CI"],
        alpha=0.2,
        label="95% Confidence Interval"
    )

    plt.axvline(test.index[0], linestyle="--", label="Forecast Start")
    plt.title(f"Sequential SARIMA + LSTM Hybrid Forecast – {book_name}")
    plt.xlabel("Date")
    plt.ylabel("Sales Volume")
    plt.legend()
    plt.show()

    return {
        "book_name": book_name,
        "train": sarima_results["train"],
        "test": test,
        "series": series,
        "forecast": hybrid_forecast,
        "sarima_forecast": sarima_forecast,
        "residual_forecast": residual_forecast,
        "hybrid_forecast": hybrid_forecast,
        "confidence_interval": confidence_interval,
        "mae": mae,
        "mape": mape,
        "preview": preview
    }
```

## Building a parallel SARIMA–LSTM hybrid:

In the sequential hybrid, SARIMA produces the main forecast first and the LSTM is then trained to forecast SARIMA’s residual errors as a correction layer. In the parallel hybrid, SARIMA and LSTM generate separate forecasts independently, and the final forecast is created by combining them using an optimised weighting scheme.

```python
# Generate SARIMA forecast on original scale.

def get_sarima_forecast_original_scale(sarima_results, horizon=32):
    test = sarima_results["test"]
    model = sarima_results["model"]
    use_log = sarima_results.get("use_log", False)

    forecast_values = model.predict(
        n_periods=horizon
    )

    if use_log:
        forecast_values = np.expm1(forecast_values)

    forecast = pd.Series(
        forecast_values,
        index=test.index,
        name="SARIMA Forecast"
    )

    return forecast
```

```python
# Create weighted-average parallel hybrid forecast.

def create_parallel_hybrid_forecast(
    sarima_forecast,
    lstm_forecast,
    sarima_weight=0.5
):
    lstm_weight = 1 - sarima_weight

    hybrid_forecast = (
        sarima_weight * sarima_forecast
        + lstm_weight * lstm_forecast
    )

    hybrid_forecast = hybrid_forecast.clip(lower=0)
    hybrid_forecast.name = "Parallel Hybrid Forecast"

    return hybrid_forecast
```

This function tests different SARIMA/LSTM blend ratios, calculates the resulting MAE and MAPE for each weighted hybrid forecast, and selects the weighting scheme with the lowest MAE as the final parallel hybrid combination.

```python
# Optimise weighting scheme.

def optimise_parallel_hybrid_weight(
    test,
    sarima_forecast,
    lstm_forecast,
    weights=np.arange(0, 1.01, 0.05)
):
    results = []

    for w in weights:
        hybrid_forecast = create_parallel_hybrid_forecast(
            sarima_forecast=sarima_forecast,
            lstm_forecast=lstm_forecast,
            sarima_weight=w
        )

        mae = mean_absolute_error(test, hybrid_forecast)
        mape = mean_absolute_percentage_error(test, hybrid_forecast)

        results.append({
            "sarima_weight": w,
            "lstm_weight": 1 - w,
            "mae": mae,
            "mape": mape
        })

    weight_results = pd.DataFrame(results).sort_values("mae")
    best_weight = weight_results.iloc[0]

    return best_weight, weight_results
```

```python
# Evaluate and plot parallel hybrid forecast.

def evaluate_parallel_hybrid(
    sarima_results,
    sarima_forecast,
    lstm_forecast,
    hybrid_forecast,
    best_weight,
    weight_results,
    book_name
):
    test = sarima_results["test"]
    series = sarima_results["series"]

    mae = mean_absolute_error(test, hybrid_forecast)
    mape = mean_absolute_percentage_error(test, hybrid_forecast)

    confidence_interval = create_hybrid_confidence_interval(
        forecast=hybrid_forecast,
        residuals=test - hybrid_forecast
    )

    print("\n" + "=" * 70)
    print(f"PARALLEL SARIMA + LSTM HYBRID – {book_name.upper()}")
    print("=" * 70)

    print("\nBest weighting scheme:")
    print(f"SARIMA weight: {best_weight['sarima_weight']:.2f}")
    print(f"LSTM weight:   {best_weight['lstm_weight']:.2f}")

    print("\nFinal holdout performance:")
    print(f"MAE:  {mae:.2f}")
    print(f"MAPE: {mape * 100:.2f}%")

    preview = pd.DataFrame({
        "Actual": test,
        "SARIMA Forecast": sarima_forecast,
        "LSTM Forecast": lstm_forecast,
        "Parallel Hybrid Forecast": hybrid_forecast,
        "Lower CI": confidence_interval["Lower CI"],
        "Upper CI": confidence_interval["Upper CI"],
        "Hybrid Error": test - hybrid_forecast,
        "Absolute Error": (test - hybrid_forecast).abs()
    })

    print("\nForecast preview:")
    display(preview.head(8).round(2))

    print("\nTop weighting results:")
    display(weight_results.head(10).round(4))

    plt.figure(figsize=(14, 5))
    series.iloc[-150:].plot(label="Observed")
    sarima_forecast.plot(label="SARIMA Forecast", linestyle="--")
    lstm_forecast.plot(label="LSTM Forecast", linestyle=":")
    hybrid_forecast.plot(label="Parallel Hybrid Forecast")

    plt.fill_between(
        confidence_interval.index,
        confidence_interval["Lower CI"],
        confidence_interval["Upper CI"],
        alpha=0.2,
        label="95% Confidence Interval"
    )

    plt.axvline(test.index[0], linestyle="--", label="Forecast Start")
    plt.title(f"Parallel SARIMA + LSTM Hybrid Forecast – {book_name}")
    plt.xlabel("Date")
    plt.ylabel("Sales Volume")
    plt.legend()
    plt.show()

    return {
        "book_name": book_name,
        "train": sarima_results["train"],
        "test": test,
        "series": series,
        "forecast": hybrid_forecast,
        "sarima_forecast": sarima_forecast,
        "lstm_forecast": lstm_forecast,
        "hybrid_forecast": hybrid_forecast,
        "best_weight": best_weight,
        "weight_results": weight_results,
        "confidence_interval": confidence_interval,
        "mae": mae,
        "mape": mape,
        "preview": preview
    }
```

```python
# Full parallel hybrid workflow.

def run_parallel_hybrid_sarima_lstm(
    sarima_results,
    lstm_results,
    book_name,
    horizon=32
):
    test = sarima_results["test"]

    sarima_forecast = get_sarima_forecast_original_scale(
        sarima_results=sarima_results,
        horizon=horizon
    )

    lstm_forecast = lstm_results["forecast"].copy()
    lstm_forecast = lstm_forecast.reindex(test.index)
    lstm_forecast.name = "LSTM Forecast"

    best_weight, weight_results = optimise_parallel_hybrid_weight(
        test=test,
        sarima_forecast=sarima_forecast,
        lstm_forecast=lstm_forecast
    )

    hybrid_forecast = create_parallel_hybrid_forecast(
        sarima_forecast=sarima_forecast,
        lstm_forecast=lstm_forecast,
        sarima_weight=best_weight["sarima_weight"]
    )

    results = evaluate_parallel_hybrid(
        sarima_results=sarima_results,
        sarima_forecast=sarima_forecast,
        lstm_forecast=lstm_forecast,
        hybrid_forecast=hybrid_forecast,
        best_weight=best_weight,
        weight_results=weight_results,
        book_name=book_name
    )

    return results
```

## Interpolating COVID-19 Data Gaps
For experimental purposes, we replace the artificial zeros introduced during the 2020-2021 reporting gaps with interpolated values. We will then rebuild the sequential SARIMA-LSTM hybrid model to evaluate whether this smoother baseline improves forecasting performance.

```python
# Function to interpolate zero-sales gaps.

def interpolate_covid_gaps(series, book_name):
    y = series.copy()

    # Replace 0s with NaN to allow interpolation.
    y_na = y.replace(0, np.nan)

    # Interpolate using time-based method.
    y_interpolated = y_na.interpolate(method="time")

    # Visualise the interpolation.
    plt.figure(figsize=(14, 5))
    plt.plot(y.index, y, label="Original (with Zeros)", alpha=0.5, color='gray')
    plt.plot(y_interpolated.index, y_interpolated, label="Interpolated", color='blue', alpha=0.8, linestyle='--')
    plt.title(f"COVID-19 Gap Interpolation – {book_name}")
    plt.xlabel("Date")
    plt.ylabel("Sales Volume")
    plt.legend()
    plt.show()

    return y_interpolated

# Apply to both books.

vhc_interpolated_series = interpolate_covid_gaps(
    very_hungry_caterpillar_series,
    "The Very Hungry Caterpillar"
)

alchemist_interpolated_series = interpolate_covid_gaps(
    alchemist_series,
    "The Alchemist"
)
```

The interpolation replaces the COVID-period zero sales gaps with estimated values that follow the surrounding trend, preventing the model from treating missing/disrupted weeks as genuine collapses in demand. This creates a smoother and more realistic time series, preserving the broader upward pattern while reducing artificial volatility caused by data gaps rather than true sales behaviour.

# Monthly Forecasting & Frequency Comparison:

This section tests whether lower-frequency monthly aggregation improves forecast stability and relative accuracy versus weekly models. Weekly sales are first aggregated into monthly totals, then XGBoost and Auto ARIMA are trained using all monthly observations up to, but excluding, the final 8 months. The final 8 months are held out for evaluation only.

```python
# Function to aggregate weekly sales into monthly sales.

def aggregate_weekly_to_monthly(series, book_name):
    y = series.dropna().copy()
    y = y.sort_index()

    # Ensure numeric values.
    y = pd.to_numeric(y, errors="coerce").dropna()

    # Convert weekly End_Date index into monthly periods, then sum monthly sales.
    monthly_sales = (
        y
        .groupby(y.index.to_period("M"))
        .sum()
    )

    # Convert PeriodIndex back to timestamp for plotting/model compatibility.
    monthly_sales.index = monthly_sales.index.to_timestamp()
    monthly_sales.name = "Monthly Sales"

    print("\n" + "=" * 70)
    print(f"MONTHLY SALES AGGREGATION – {book_name.upper()}")
    print("=" * 70)
    print(f"Start date     : {monthly_sales.index.min()}")
    print(f"End date       : {monthly_sales.index.max()}")
    print(f"Rows/months    : {len(monthly_sales)}")
    print(f"Missing values : {monthly_sales.isna().sum()}")
    print(f"Min value      : {monthly_sales.min():.2f}")
    print(f"Max value      : {monthly_sales.max():.2f}")

    display(monthly_sales.head().to_frame())
    display(monthly_sales.tail().to_frame())

    plt.figure(figsize=(14, 5))

    monthly_sales.plot(
        label="Monthly Sales"
    )

    plt.title(
        f"Monthly Sales Series – {book_name}"
    )

    plt.xlabel("Date")
    plt.ylabel("Monthly Sales")
    plt.legend()
    plt.grid(alpha=0.3)

    plt.show()

    return monthly_sales
```

```python
# Aggregate weekly sales into monthly sales.

vhc_monthly_sales = aggregate_weekly_to_monthly(
    very_hungry_caterpillar_series,
    "The Very Hungry Caterpillar"
)
```

* Monthly sales range from a minimum of 0 to a maximum of 13,949, showing that even after aggregation the series remains highly volatile.
* The chart shows a clear long-term increase in monthly sales levels, with sales generally much higher from 2021 onwards than in the earlier years.
* The aggregation smooths some of the week-to-week noise, making broader demand patterns easier to see.
* However, the monthly series still contains major spikes and disruption periods, including the COVID-era zero values and large peaks in later years.
* Compared with weekly modelling, monthly resampling reduces the amount of training data substantially, which can make models less able to learn detailed seasonal patterns.
* Overall, the monthly version is useful for analysing broader sales trends, but it sacrifices weekly granularity and may be less effective for capturing sharp short-term movements such as holiday spikes.

```python
# Create monthly train/test split.

def create_monthly_train_test_split(
    monthly_series,
    book_name,
    horizon=8
):
    monthly_series = monthly_series.dropna().copy()

    train = monthly_series.iloc[:-horizon]
    test = monthly_series.iloc[-horizon:]

    print("\n" + "=" * 70)
    print(f"MONTHLY TRAIN / TEST SPLIT – {book_name.upper()}")
    print("=" * 70)

    print(f"Total observations : {len(monthly_series)}")
    print(f"Training months    : {len(train)}")
    print(f"Test months        : {len(test)}")

    print("\nTraining period:")
    print(f"{train.index.min()} to {train.index.max()}")

    print("\nTest period:")
    print(f"{test.index.min()} to {test.index.max()}")

    return {
        "series": monthly_series,
        "train": train,
        "test": test,
        "horizon": horizon
    }
```
